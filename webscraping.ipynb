```python
def parsearReceta(html_doc_path):
    with open(html_doc_path, "r", encoding="utf-8") as f:
        contenido_html = f.read()
    soup = BeautifulSoup(contenido_html, 'html.parser')
    
    score = soup.find("div", {"id": "mm-recipes-review-bar__rating_1-0"}).string
    description = soup.find("p", {"class": "article-subheading text-utility-300"}).string
    prepTime = soup.find("div", {"class": "mm-recipes-details__value"}).string
    title = soup.find("meta", {"property": "og:title"}).string
    return{"score":score,"prepTime":prepTime,"description":description,"title":title}
```

```python
htmlparseado = parsearReceta("/kaggle/input/unstuffed-cabbage-roll-recipe-2-0/Unstuffed Cabbage Roll Recipe.html")
htmlparseado
```

```python
import requests
from bs4 import BeautifulSoup
import time

url_indice = "https://www.allrecipes.com/" # Cambia por la URL real
headers = {'User-Agent': 'Mozilla/5.0'}

response = requests.get(url_indice, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Buscamos todos los enlaces a recetas. 
# NOTA: Debes inspeccionar la página para ver qué clase tienen los enlaces.
# Ejemplo común: <a class="card__title" href="...">
links = soup.find_all('a', class_='mntl-card-list-items') 

urls_recetas = []
for link in links:
    href = link.get('href')
    if href:
        urls_recetas.append(href)

print(f"Se encontraron {len(urls_recetas)} recetas.")
urls_recetas
```

```python
# Ejemplo para leer un sitemap
response = requests.get("https://www.allrecipes.com/")
soup = BeautifulSoup(response.content, 'xml') # Nota el uso de 'xml'

todas_las_urls = [loc.text for loc in soup.find_all('loc')]
print(f"¡Brutal! Encontré {len(todas_las_urls)} recetas directas.")
```
